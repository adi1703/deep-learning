### What are n-grams?
N-grams are continuous sequences of words or symbols or tokens in a document. In technical terms, they can be defined as the neighbouring sequences of items in a document. They come into play when we deal with text data in NLP(Natural Language Processing) tasks.

#### An example of n-grams
Let’s understand n-grams practically with the help of the following sentence:

image.png

From the table above, it’s clear that unigram means taking only one word at a time, bigram means taking two words at a time and trigram means taking three words at a time.

```python
import nltk
from nltk.util import ngrams
samplText='CloudyMl is an EduTech Company'
```

```python
nltk.download('punkt_tab')
NGRAMS=ngrams(sequence=nltk.word_tokenize(samplText), n=1)
#usning for loop for printing the values.
for grams in NGRAMS:
    print(grams)
```

```python
#we will use ngrams function from nltk
#NGRAMS=ngrams(sequence=nltk.word_tokenize(samplText), n=2)
samplText='CloudyMl is an EduTech Company'
NGRAMS=ngrams(sequence=nltk.word_tokenize(samplText), n=2)
for grams in NGRAMS:
    print(grams)
```

```python
samplText='CloudyMl is an EduTech Company'
NGRAMS=ngrams(sequence=nltk.word_tokenize(samplText), n=3)
for grams in NGRAMS:
    print(grams)
```
